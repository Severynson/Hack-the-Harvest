```python
import os, zipfile

MAX_MB = 50                                   # skip giant data files so the upload doesn't fail
out = "keeboh_all.zip"
skipped = []

with zipfile.ZipFile(out, "w", zipfile.ZIP_DEFLATED) as z:
    for root, dirs, files in os.walk("."):
        dirs[:] = [d for d in dirs if d != ".ipynb_checkpoints" and not d.startswith(".")]
        for f in files:
            path = os.path.join(root, f)
            if f == out:
                continue
            size_mb = os.path.getsize(path) / 1e6
            if size_mb > MAX_MB:
                skipped.append(f"{path} ({size_mb:.0f} MB)")
                continue
            z.write(path)

print(f"Zip size: {os.path.getsize(out)/1e6:.1f} MB")
print("Skipped (too big):", skipped or "none")
```
